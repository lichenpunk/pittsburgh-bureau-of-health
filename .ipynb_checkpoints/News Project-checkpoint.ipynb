```python
import requests, time, json
BASE = "https://www.loc.gov/collections/chronicling-america/"
params = {
    "q": "filthy",
    "dates": "1891/1893",
    "fo": "json",
    "c": 300,
    "sp": 1,
}
results = []
r = requests.get(BASE, params=params, timeout=30)
r.raise_for_status()
data = r.json()
items = data.get("results", [])
results.extend(items)
print(results)
with open('filthydata.json', 'w') as f:
    json.dump(data, f)
```
